#### **5. Let's apply 'deepface' to YouTube videos.**

```python
from deepface import DeepFace
import pandas as pd

def extract_emotion_info(thumbnail_url):
    try:
        # Analyze emotions in the image
        results = DeepFace.analyze(img_path=str(thumbnail_url), actions=['emotion'])

        # DeepFace sometimes returns a single dict or a list — handle both
        if isinstance(results, dict):
            results = [results]

        # Extract info for each face detected
        emotions_data = []
        for face_result in results:
            dominant = face_result['dominant_emotion']
            proportions = face_result['emotion']  # dictionary of all emotion scores
            emotions_data.append({
                'dominant_emotion': dominant,
                'emotion_proportions': proportions
            })

        return emotions_data

    except Exception as e:
        print(f"Error analyzing thumbnail: {e}")
        return []

# Apply to your DataFrame
trump['emotion_info'] = trump['thumbnail_maxres'].apply(extract_emotion_info)

# If you want separate columns:
trump['deepface_dominant_emotion'] = trump['emotion_info'].apply(lambda x: x[0]['dominant_emotion'] if x else None)
trump['emotion_proportions'] = trump['emotion_info'].apply(lambda x: x[0]['emotion_proportions'] if x else None)
```

#### **6. Let's make each emotion as a category.**

```python
# Explode the emotion_proportions dictionary into separate columns
emotions_df = trump['emotion_proportions'].apply(pd.Series)

# Add the prefix "deepface_" to each emotion column
emotions_df = emotions_df.add_prefix('deepface_')

# Concatenate back to your main DataFrame
trump = pd.concat([trump, emotions_df], axis=1)

#drop emotion_proportions and emotion_info columns
trump = trump.drop(columns=['emotion_info', 'emotion_proportions'])

# Print first 10 rows of trump dataset again
trump.head(10)
```
